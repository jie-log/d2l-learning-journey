```python
# 创建浮点型张量（因为第一个元素是1.0，会自动转为float32）
x = torch.tensor([1.0,2,4,8])
# 创建整型张量（默认int64）
y = torch.tensor([2,2,2,2])
print(x + y)
print(x - y)
print(x * y)  # 逐元素乘法（哈达玛积，非矩阵乘法）
print(x / y)  # 逐元素真除法（结果总是浮点数，即使两个整数相除）
print(x % y)  # 逐元素取模（余数）,结果符号与除数相同
print(x ** y) # 逐元素幂运算：x 的 y 次方
# 结果带小数点是因为 运算结果的张量是浮点类型
```

**在 PyTorch 中，`x += y` 并不完全等价于 `x = x + y`**，两者行为不同。
- **`x = x + y`**
  先计算 `x + y`（返回**新的张量**），然后把新张量赋值给变量 `x`。原来的张量没有被修改，只是变量 `x` 换了个“指向”。
  → `id(x)` 会改变 → 结果是 `False`

- **`x += y`**
  调用的是 `x.__iadd__(y)` 方法（即**原地加法**）。对于 PyTorch 张量，这个方法会在**不新建张量**的前提下，直接修改 `x` 原来的内存内容。
  → `id(x)` 保持不变 → 结果是 `True`

### Python 的通用规则
- 对于**不可变类型**（如 `int`, `str`, `tuple`）：`x += y` 会先计算新对象再重新赋值，等价于 `x = x + y`。
- 对于**可变类型**（如 `list`, `dict`, `torch.Tensor`）：`x += y` 通常会设计为**原地操作**（修改原对象），而 `x = x + y` 总是创建新对象。
