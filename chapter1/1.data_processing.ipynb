```python
import torch
from pandas.core.arrays.arrow import dtype

x = torch.arange(12)
x
```

```python
x.shape # 查看张量的形状（维度大小）
# 因为x是一维的，有12个元素，所以shape是 torch.Size([12])
```

```python
x.numel() #元素的种数or返回张量中元素的总个数（与shape中各维度乘积相同）
```

```python
x.reshape(3,4)  # 输出一个形状为(3,4)的二维张量
```

```python
torch.zeros((2,3,4)) #深度2 长3 宽4 的三维张量
```

```python
torch.ones((2,3,4))
```

```python
torch.tensor([[2,1,4,3],
              [4,5,6,7],
              [8,9,10,11]])
# 使用嵌套列表直接创建张量
```

```python
torch.tensor([[[2,1,4,3],
              [4,5,6,7],
              [8,9,10,11]]]).shape
# 理解：这是一个1张3×4的矩阵的三维张量。相当于在一个深度（batch/depth）为1的容器里放了一张3行4列的表格。
```

```python
# 创建浮点型张量（因为第一个元素是1.0，会自动转为float32）
x = torch.tensor([1.0,2,4,8])
# 创建整型张量（默认int64）
y = torch.tensor([2,2,2,2])
print(x + y)
print(x - y)
print(x * y)  # 逐元素乘法（哈达玛积，非矩阵乘法）
print(x / y)  # 逐元素真除法（结果总是浮点数，即使两个整数相除）
print(x % y)  # 逐元素取模（余数）,结果符号与除数相同
print(x ** y) # 逐元素幂运算：x 的 y 次方
# 结果带小数点是因为 运算结果的张量是浮点类型
```

```python
torch.exp(x)
```

```python
X = torch.arange(12,dtype=torch.float32).reshape((3,4))
# 外层括号：是函数 reshape() 的调用括号，必不可少。
# 内层括号：是 元组 (3,4) 的括号，因为 reshape 需要一个表示形状的参数，这里我们传入一个元组 (3,4) 作为参数。
Y = torch.tensor(([2.0,1,4,3],
                  [4,5,6,7],
                  [8,9,10,11]))
torch.cat((X,Y),dim = 0), torch.cat((X,Y),dim = 1)
# 输出带(是因为用了逗号把两个torch.cat的结果连在了一起,形成一个元组（tuple）
```

```python
X == Y
```

```python
X.sum()
```

```python
# 广播机制：当两个张量形状不同时，PyTorch 会自动尝试通过“扩展”某些维度，使它们形状兼容，从而执行逐元素运算（如加法、乘法等）。这个过程不需要复制数据，内存高效。
a = torch.arange(3).reshape((3,1))
b = torch.arange(2).reshape((1,2))
a,b
```

```python
a + b
```
