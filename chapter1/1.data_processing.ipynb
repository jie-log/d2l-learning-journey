```python
import torch
x = torch.arange(12)
x
```

```python
x.shape # 查看张量的形状（维度大小）
# 因为x是一维的，有12个元素，所以shape是 torch.Size([12])
```

```python
x.numel() #元素的种数or返回张量中元素的总个数（与shape中各维度乘积相同）
```

```python
x.reshape(3,4)  # 输出一个形状为(3,4)的二维张量
```

```python
torch.zeros((2,3,4)) #深度2 长3 宽4 的三维张量
```

```python
torch.ones((2,3,4))
```

```python
torch.tensor([[2,1,4,3],
              [4,5,6,7],
              [8,9,10,11]])
# 使用嵌套列表直接创建张量
```

```python
torch.tensor([[[2,1,4,3],
              [4,5,6,7],
              [8,9,10,11]]]).shape
# 理解：这是一个1张3×4的矩阵的三维张量。相当于在一个深度（batch/depth）为1的容器里放了一张3行4列的表格。
```

```python
# 创建浮点型张量（因为第一个元素是1.0，会自动转为float32）
x = torch.tensor([1.0,2,4,8])
# 创建整型张量（默认int64）
y = torch.tensor([2,2,2,2])
print(x + y)
print(x - y)
print(x * y)  # 逐元素乘法（哈达玛积，非矩阵乘法）
print(x / y)  # 逐元素真除法（结果总是浮点数，即使两个整数相除）
print(x % y)  # 逐元素取模（余数）,结果符号与除数相同
print(x ** y) # 逐元素幂运算：x 的 y 次方
```
