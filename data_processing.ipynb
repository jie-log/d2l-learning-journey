```python
import torch
x = torch.arange(12)
x
```

```python
x.shape # 查看张量的形状（维度大小）
# 因为x是一维的，有12个元素，所以shape是 torch.Size([12])
```

```python
x.numel() #元素的种数or返回张量中元素的总个数（与shape中各维度乘积相同）
```

```python
x.reshape(3,4)  # 输出一个形状为(3,4)的二维张量
```

```python
torch.zeros((2,3,4)) #深度2 长3 宽4 的三维张量
```

```python
torch.ones((2,3,4))
```
